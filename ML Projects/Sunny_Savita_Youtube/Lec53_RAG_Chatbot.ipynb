```python
import os
import warnings
warnings.filterwarnings("ignore")
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
from langchain_text_splitters import RecursiveCharacterTextSplitter, CharacterTextSplitter
from langchain_openai import ChatOpenAI
from langchain.schema.runnable import RunnablePassthrough, RunnableLambda
from langchain.schema.output_parser import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_community.tools import TavilySearchResults
from langchain_core.messages import HumanMessage, SystemMessage, trim_messages
from langgraph.prebuilt import create_react_agent, ToolNode
from langgraph.checkpoint.memory import MemorySaver
from langgraph.graph import END, StateGraph, MessagesState
from langchain_core.tools import tool
from typing import Literal
from google.colab import userdata
import bs4
from langchain.vectorstores import Chroma
import chromadb
from langchain_community.document_loaders import WebBaseLoader
from langchain_core.messages import HumanMessage, AIMessage
```

```python
OPENAI_API_TOKEN=userdata.get('OPENAI_API_KEY')
LANGCHAIN_API_KEY = userdata.get('LANGCHAIN_API_KEY')
LANGSMITH_TRACING = userdata.get('LANGSMITH_TRACING')
LANGSMITH_ENDPOINT = userdata.get('LANGSMITH_ENDPOINT')
LANGSMITH_PROJECT = userdata.get('LANGSMITH_PROJECT')
os.environ["OPENAI_API_KEY"] = OPENAI_API_TOKEN
os.environ["LANGCHAIN_API_KEY"] = LANGCHAIN_API_KEY
os.environ["LANGSMITH_TRACING"] = LANGSMITH_TRACING
os.environ["LANGSMITH_ENDPOINT"] = LANGSMITH_ENDPOINT
os.environ["LANGSMITH_PROJECT"] = LANGSMITH_PROJECT
embeddings = OpenAIEmbeddings()
llm = ChatOpenAI()
parser = StrOutputParser()
```

```python
loader = WebBaseLoader(web_paths=("https://lilianweng.github.io/posts/2023-06-23-agent/",),bs_kwargs=dict(parse_only=bs4.SoupStrainer(class_=("post-content", "post-title", "post-header"))),)
doc = loader.load()
doc
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
splits = text_splitter.split_documents(doc)
persist_directory = "db"
os.makedirs(persist_directory, exist_ok=True)

vectorstore = Chroma.from_documents(splits, embeddings, persist_directory=persist_directory)
retriever = vectorstore.as_retriever()
```

```python
system_prompt = (
    "You are an assistant for question-answering tasks. "
    "Use the following pieces of retrieved context to answer the question "
    "If you don't know the answer, say that you don't know."
    "Use three sentences maximum and keep the answer concise."
    "\n\n"
    "{context}"
)
```

```python
chat_prompt = ChatPromptTemplate.from_messages(
    [
        ("system", system_prompt),
        ("human", "{input}"),
    ]
)
```

```python
from langchain.chains.combine_documents import create_stuff_documents_chain
from langchain.chains import create_retrieval_chain
```

```python
question_answering_chain=create_stuff_documents_chain(llm, chat_prompt)
rag_chain = create_retrieval_chain(retriever, question_answering_chain)
response = rag_chain.invoke({"context":"MRKL stands for Modular Reasoning, Knowledge and Language,which is a neuro-symbolic architecture for autonomous agents. It consists of a collection of expert modules where a general-purpose LLM routes inquiries to the best suitable expert module, which can be neural (e.g. deep learning models) or symbolic (e.g. math calculator, weather API).", "input":"what is MRKL?"})
response["answer"]
```

Adding Memory History

```python
from langchain.chains import create_history_aware_retriever
```

```python
retriever_prompt = (
    "Given a chat history and the latest user question which might reference context in the chat history,"
    "formulate a standalone question which can be understood without the chat history."
    "Do NOT answer the question, just reformulate it if needed and otherwise return it as is."
)
```

```python
contextualize_q_prompt  = ChatPromptTemplate.from_messages( # Create Contextualized Query Prompt
    [
        ("system", retriever_prompt),
        MessagesPlaceholder(variable_name="chat_history"),
        ("human", "{input}"),


     ]
)
```

```python
history_aware_retriever = create_history_aware_retriever(llm,retriever,contextualize_q_prompt)
```

```python
qa_prompt = ChatPromptTemplate.from_messages( # Main Prompt with the contextualized query prompt to provide better context
    [
        ("system", system_prompt),
        MessagesPlaceholder("chat_history"),
        ("human", "{input}"),
    ]
)
```

```python
question_answer_chain = create_stuff_documents_chain(llm, qa_prompt)
rag_chain = create_retrieval_chain(history_aware_retriever, question_answer_chain)
```

```python
chat_history = []
```

```python
question1 = "what is Task Decomposition?"
message1= rag_chain.invoke({"input": question1, "chat_history": chat_history})
message1["answer"]
```

```python
chat_history.extend(
    [
        HumanMessage(content=question1),
        AIMessage(content=message1["answer"]),
    ]
)
chat_history
```

```python
second_question = "What are common ways of doing it?"
message2 = rag_chain.invoke({"input": second_question, "chat_history": chat_history})

print(message2["answer"])
```

Method 2 (With working contextualized query)

```python
from langchain.chains import LLMChain
```

```python
reformulation_chain = LLMChain(
    llm=llm,
    prompt=contextualize_q_prompt
)
```

```python
def reformulate_and_retrieve(inputs):
    # Get the reformulated query
    reformulated = reformulation_chain.invoke({
        "input": inputs["input"],
        "chat_history": inputs["chat_history"]
    })
    print("\nOriginal Question:", inputs["input"])
    print("\nReformulated Question:", reformulated["text"])

    # Get relevant documents
    docs = retriever.get_relevant_documents(reformulated["text"])
    return docs
```

```python
retrieval_chain = RunnablePassthrough.assign(
    context=RunnableLambda(reformulate_and_retrieve)
)
```

```python
question_answer_chain = create_stuff_documents_chain(llm, qa_prompt)
rag_chain = retrieval_chain.assign(answer=question_answer_chain)
```

```python
chat_history = []
```

```python
# First question
question1 = "what is Task Decomposition?"
print("\n=== First Question ===")
message1 = rag_chain.invoke({
    "input": question1,
    "chat_history": chat_history
})
```

```python
chat_history.extend([
    HumanMessage(content=question1),
    AIMessage(content=message1["answer"])
])
```

```python
second_question = "What are common ways of doing it?"
print("\n=== Follow-up Question ===")
message2 = rag_chain.invoke({
    "input": second_question,
    "chat_history": chat_history
})
```

```python
message2
```

```python
reformulated = reformulation_chain.invoke({
    "input": second_question,
    "chat_history": chat_history
})
print("\nReformulated Query from message2:", reformulated["text"])
```
