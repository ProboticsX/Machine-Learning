```python
from pinecone import Pinecone

pc = Pinecone(api_key=PINECONE_API_KEY)
index = pc.Index("mongo")
```

```python
index
```

```python
def get_result(query, similar_result = 3):
  vector = embeddings.embed_query(query)
  result = index.query(vector = vector, top_k = similar_result)
  return result
```

```python
query="what is the best comedy movie to watch and why?"
```

```python
result=get_result(query)
```

```python
result
```

```python
from bson.objectid import ObjectId
mylist = []
for i in range(len(result["matches"])):
  value = result["matches"][i]["id"]
  mylist.append(collection.find_one({"_id": ObjectId(value)}))
```

```python
mylist
```

```python
combined_information = ""
for i in range(len(mylist)):
  combined_information += f"Title: {mylist[i].get('title', 'N/A')}, Plot: {mylist[i].get('fullplot', 'N/A')}\n"

print(combined_information)
```

```python
template = """
Query: {query}
Continue to answer the query by using the Search Results:
{combined_information}
"""
```

```python
prompt = ChatPromptTemplate.from_template(template)
```

```python
model = ChatOpenAI(model="gpt-4o-mini")
```

```python
rag_chain = (
    RunnablePassthrough()
    | RunnableLambda(lambda x: {"query": x, "combined_information": combined_information})
    | prompt
    | model
    | StrOutputParser()
)
```

```python
result = rag_chain.invoke(query)
```
