```python
system_prompt = "You are an LLM Python Code Tutor. \
You will be given a code block in Python.\n"
system_prompt += "You should explain what does the code does and the reasoning behind it"
```

```python
def get_user_prompt(code):
    user_prompt = f"You are looking at a code block:\n"
    user_prompt += f"{code}"
    return user_prompt
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
get_user_prompt('yield from {book.get("author") for book in books if book.get("author")')
```

```python
def call_llm_tutor(code):
    response = openai.chat.completions.create(
        model=MODEL_GPT,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": get_user_prompt(code)}
          ],
    )
    result = response.choices[0].message.content
    display(Markdown(result))
```
